```python
import pandas as pd
import numpy as np

#generate fake data
np.random.seed(0)
```

NOMINAL DATA

```python
num_samples = 1000
colors = ['red','blue', 'green','yelow', 'black']
nominal_data = np.random.choice(colors, num_samples)

#create the dataset
# Fav Color = > Nominal data
df = pd.DataFrame(nominal_data, columns=['Fav color'])
df
```

ORDINAL DATA

```python
education_qualification = ['12th', 'B.tech', 'M.Tech', 'phD']
ordinal_data = np.random.choice(education_qualification, num_samples)
df['Educational Qualification'] = ordinal_data
df
```

Interval Data

```python
interval_data = np.random.randint(0,100,num_samples)
df['Temperature(Fahrenheit)'] = pd.DataFrame(interval_data)
df
```

RATIO DATA

```python
# INCOME IN DOLLARS
ratio_data = np.random.randint(1,100,num_samples) * 1000
df['Income'] = ratio_data
df
```

```python
#DISCRETE DATA
discrete_data = np.random.randint(1, 10, num_samples)
df['Num of Children'] = discrete_data
df
```

Continuous Data

```python
continuous_data = np.random.rand(num_samples) * 100
df['Weight'] = continuous_data
df
```

RANDOM SAMPLING

Take 100 samples from the above dataframe via random sampling

```python
random_sample = df.sample(n=100, random_state=42)
random_sample
```

Stratified Sampling On population Data by Region

```python
population_data = pd.read_csv('/content/Population_Survey_Data.csv')
manufacturing_data = pd.read_csv('/content/Manufacturing_Data.csv')
```

```python
stratified_sample = population_data.groupby('Region').apply(lambda x: x.sample(min(len(x), 25)))
stratified_sample
stratified_sample.to_csv('stratified_sample.csv', index=False)
```

CLUSTER SAMPLING

```python
manufacturing_data = pd.read_csv('/content/Manufacturing_Data.csv')
manufacturing_data
```

```python
clusters = manufacturing_data['BatchNumber'].unique()
print(clusters)
```

```python
selected_clusters = np.random.choice(clusters, size=5, replace=False)
print(selected_clusters)
```

```python
systematic_sample = population_data.iloc[::10, :]
print(systematic_sample)
```

```python
mean_income = population_data['Income'].mean()
median_income = population_data['Income'].median()
mode_income = population_data['Income'].mode()[0]
print("Mean Income:", mean_income)
print("Median Income:", median_income)
print("Mode Income:", mode_income)
```
