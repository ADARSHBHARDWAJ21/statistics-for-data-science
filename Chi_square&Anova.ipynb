```python
#import all the required libraries
import pandas as pd
import numpy as np
import scipy.stats as stats
```

Imagine youare data analyst working for a retail company , The company wants to know if customer prefrence for different age groups. To invastigate this you conduct a survey and collect the foloowing data.


```python
observed_data = np.array([[50, 30 , 20],
                         [30,40,30],
                         [20,30,50]])
```

Perform chi Square test

```python
chi2_test, p_val, dof, expected_data= stats.chi2_contingency(observed_data)
```

```python
print(chi2_test)
```

```python
print(expected_data)
```

```python
print(p_val)
```

```python
print(dof)
```

Final conclusion

```python
alpha = 0.05

if p_val < alpha:
  print('Reject the null hypothesis')
else:
  print("Fail to reject the null hypothesis")
```
